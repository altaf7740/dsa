## Factorial of Number

```python
def factorial(number):
    if number in (0,1):
        return number

    small_step = factorial(number-1) # from PMI
    return small_step * number


print(factorial(5))
print(factorial(4))
print(factorial(1))
print(factorial(0))
```

## Sum of natural number

```python
def sum_natural_number(number):
    if number in (0,1):
        return number

    small_step = sum_natural_number(number-1) # from PMI
    return number + small_step


print(sum_natural_number(5))
print(sum_natural_number(4))
print(sum_natural_number(1))
print(sum_natural_number(0))
```

## Number of digits

```python
def total_digits(number):
    if number in range(10):
        return 1

    small_step = total_digits(int(number/10))
    return small_step + 1


print(total_digits(12345))
print(total_digits(125))
print(total_digits(1))
print(total_digits(0))
```

## Sum of digits

```python
def sum_of_digit(number):
    if number in range(10):
        return number

    small_step = sum_of_digit(int(number/10))
    return small_step + number%10

print(sum_of_digit(12345))
print(sum_of_digit(125))
print(sum_of_digit(1))
print(sum_of_digit(0))
```

## Fibonacci

```python
def fibonacci(nth):
    if nth in (0,1,2):
        return nth

    last = fibonacci(nth-1)
    second_last = fibonacci(nth-2)
    nth_fibonacci = last + second_last

    return nth_fibonacci


print(fibonacci(5))
print(fibonacci(8))
print(fibonacci(1))
print(fibonacci(2))
```

## Print 1 to N

```python
def print_to_n(number):
    if number == 0:
        return number

    small_step = print_to_n(number-1)
    print(number)

print_to_n(10)
print()
print_to_n(5)
```

## Print N to 1

```python
def print_to_n(number):
    if number == 0:
        return number

    print(number)
    small_step = print_to_n(number-1)

print_to_n(10)
print()
print_to_n(5)
```

## Check if An Array is Sorted

```python
def is_sorted(array, index=None):
    if index is None:
        index = len(array) -1 
    if index <= 0:
        return True
    if array[index - 1] > array[index]:
        return False
    return is_sorted(array, index-1)
    

is_sorted([2,3,14,5,6])
```

## Sum of an Array

```python
def sum_of_array(array, index=None):
    if index is None:
        index = len(array)-1
    if index == 0:
        return array[index]
    return array[index] + sum_of_array(array, index-1)

print(sum_of_array([1,2,3,4,5]))
print(sum_of_array([3,5,2,1]))
print(sum_of_array([4]))
```
